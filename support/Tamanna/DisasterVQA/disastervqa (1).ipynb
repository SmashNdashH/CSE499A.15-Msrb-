## 2. Load DisasterVQA Dataset
Using the Hugging Face `datasets` library to load `QCRI/DisasterVQA`.

```python
from datasets import load_dataset
import json

dataset = load_dataset("QCRI/DisasterVQA", split="train")
print(f"Total QA pairs loaded: {len(dataset)}")
```

## 3. Load Abrar's Qwen2.5-VL Model
Load `AbrarAlam/disasterm3-qwen2.5vl7b-mergedFP`. We use `sdpa` for memory efficiency without needing flash-attention.

```python
import torch
from transformers import Qwen2_5_VLForConditionalGeneration, AutoProcessor

model_id = "AbrarAlam/disasterm3-qwen2.5vl7b-mergedFP"

model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    model_id, 
    torch_dtype=torch.bfloat16, 
    device_map="auto",
    attn_implementation="sdpa"
)
processor = AutoProcessor.from_pretrained(model_id)
```

## 4. Dataset Inference Loop
Run the model on the evaluation data.

```python
from datasets import load_dataset
from qwen_vl_utils import process_vision_info
from tqdm.notebook import tqdm
import json
import torch

results = []
limit = 100 
eval_dataset = dataset.select(range(min(limit, len(dataset))))

for item in tqdm(eval_dataset, desc="Running Inference"):
    try:
        image = item["image"].convert("RGB")
        question = item["question"]
        q_type = item["question_type"]
        category = item.get("crisis_info_type", "General")
        
        if q_type == "Multiple-Choice":
            choices = item.get("choices", {})
            choices_str = "\n".join([f"{k}: {v}" for k, v in choices.items() if v])
            instruction = f"{question}\nChoices:\n{choices_str}\nAnswer with the correct option letter."
            gt_list = item.get("groundtruth_answer")
            ground_truth = str(gt_list[0]) if gt_list else ""
        else:
            instruction = f"{question}\nProvide a concise and accurate answer."
            gt_list = item.get("groundtruth_answer")
            ground_truth = str(gt_list[0]) if gt_list else ""
            
        messages = [
            {
                "role": "user",
                "content": [
                    {"type": "image", "image": image},
                    {"type": "text", "text": instruction},
                ],
            }
        ]
        
        text = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        image_inputs, video_inputs = process_vision_info(messages)
    
        inputs = processor(
            text=[text],
            images=image_inputs,
            videos=video_inputs,
            padding=True,
            return_tensors="pt"
        ).to("cuda")
    
        with torch.no_grad():
            generated_ids = model.generate(**inputs, max_new_tokens=128)
            generated_ids_trimmed = [
                out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)
            ]
            output_text = processor.batch_decode(
                generated_ids_trimmed, skip_special_tokens=True, clean_up_tokenization_spaces=False
            )[0]
    
        results.append({
            "id": item["question_id"],
            "instruction": instruction,
            "ground_truth": ground_truth,
            "question_type": q_type,
            "humanitarian_category": category,
            "model_output": output_text.strip()
        })
    except Exception as e:
        print(f"Error processing item {item.get('question_id')}: {e}")
        continue

with open("vqa_predictions.jsonl", "w") as f:
    for item in results:
        f.write(json.dumps(item) + "\n")

print("Inference completed! Predictions saved to vqa_predictions.jsonl")
```

## 5. Evaluation using Gemini Judge & Exact Match

```python
import json
import pandas as pd
import time
from google import genai
from google.genai import types
from kaggle_secrets import UserSecretsClient
from tqdm.notebook import tqdm

try:
    user_secrets = UserSecretsClient()
    api_key = user_secrets.get_secret("GEMINI_API_KEY")
except Exception as e:
    print("Could not find GEMINI_API_KEY. Add it to Kaggle Secrets.")
    api_key = ""

if api_key:
    client = genai.Client(api_key=api_key)

def evaluate_open_ended(question, ground_truth, model_output):
    if not api_key: return 0
    prompt = f"""
    You are an expert disaster response evaluator.
    Question: {question}
    Ground Truth Answer: {ground_truth}
    Model Predicted Answer: {model_output}
    Is the Model Predicted Answer semantically equivalent and factually consistent with the Ground Truth Answer?
    Respond ONLY with a JSON object: {{"verdict": "CORRECT"}} or {{"verdict": "INCORRECT"}}
    """
    try:
        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=prompt,
            config=types.GenerateContentConfig(response_mime_type="application/json"),
        )
        res_json = json.loads(response.text)
        return 1 if res_json.get("verdict") == "CORRECT" else 0
    except Exception as e:
        time.sleep(2)
        return 0

try:
    with open("vqa_predictions.jsonl", "r") as f:
        results = [json.loads(line) for line in f]
except FileNotFoundError:
    print("vqa_predictions.jsonl not found. Please run the inference loop first.")
    results = []

if len(results) > 0:
    print("Evaluating answers...")
    for pred in tqdm(results, desc="Evaluating"):
        q_type = pred.get("question_type", "")
        gt = str(pred.get("ground_truth", "")).strip().lower()
        out = str(pred.get("model_output", "")).strip().lower()
        
        if q_type == "Yes/No" or q_type == "Binary":
            if ("yes" in gt and "yes" in out) or ("no" in gt and "no" in out):
                pred["is_correct"] = 1
            else:
                pred["is_correct"] = 0
        elif q_type == "Multiple-Choice":
            if len(out) > 0 and len(gt) > 0 and out[0] == gt[0]:
                pred["is_correct"] = 1
            else:
                pred["is_correct"] = 0
        else:
            pred["is_correct"] = evaluate_open_ended(pred.get("instruction", ""), gt, out)
    
    with open("vqa_predictions_evaluated.jsonl", "w") as f:
        for pred in results:
            f.write(json.dumps(pred) + "\n")
    print("✓ Saved evaluated predictions to 'vqa_predictions_evaluated.jsonl'")
    
    df = pd.DataFrame(results)
    
    if 'humanitarian_category' not in df.columns:
        df['humanitarian_category'] = 'General'
    else:
        df['humanitarian_category'] = df['humanitarian_category'].fillna('General')
    
    if 'question_type' not in df.columns:
        df['question_type'] = 'Unknown'
    else:
        df['question_type'] = df['question_type'].fillna('Unknown')
    
    summary = (
        df.groupby(['humanitarian_category', 'question_type'])['is_correct']
        .agg(
            Total='count',
            Correct='sum',
            Accuracy=lambda x: (x.sum() / x.count()) * 100 if x.count() > 0 else 0
        )
        .reset_index()
    )
    
    print("\n=== DisasterVQA Category Breakdown ===")
    print(summary.to_string(index=False))
    
    summary.to_csv("disastervqa_category_metrics.csv", index=False)
    print("\n✓ Successfully saved 'disastervqa_category_metrics.csv'!")
```
